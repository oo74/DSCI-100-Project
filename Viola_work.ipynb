### Summarize what we found:

We used the KNN classification model to explore whether age and number of hours played can predict the subscription of a player. The plot in figure 2 is the real distribution of subscribers and non-subscribers based on the player characteristics of age and played hours. It demonstrates that majority of users played very few hours, mostly are under 15 hours. But there are a few outliers with played hours greater than 50 or even 200. And the majority of players are aged between 15 and 25. We can find that subscribers tend to have more played hours and are relatively young (between 15-25). The plot in figure 3 is the KNN model to predict whether a player will subscribe the newsletter of a game. After tuning k using 5-fold cross-validation, the model achieved its highest accuracy at a specific value of k. Our KNN model achieved moderate accuracy (~74%) but struggled particularly with identifying non-subscribers due to class imbalance and possibly weak signal in the predictors. While there were slight trends, such as most subscribers are slightly younger and more active (which means have a relative long played hours).

### Is This What You Expected to Find?

Yes, mostly is what we expected to find. We expected that the player who are more younger or have more played hours tend to subscribe the newsletter of the game, as time played was assumed to reflect engagement. The KNN model prediction mostly conform to what we expect. However, the KNN model does not have a extremely high accuracy (moderate accuracy about 74%), this demonstrates that the played hours and age might not be sufficient on their own to explain subscription behavior. This deviation from our expectations highlights the complexity of user engagement and indicates that subscription decisions might depend on other factors not captured in our data.

### what impact could such findings have?

This model demonstrates the player characteristic could help the game producer to predict whether players will subscribe the newsletters to some extent, but not very accurate. This could help them to find their target consumers, improve the game experience in time, and update the content that players are interested in. And the finding could also suggest that they may need to look beyond basic activity data and consider more interaction variables to understand players' preferences.
